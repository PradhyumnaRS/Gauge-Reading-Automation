```python
import cv2
import numpy as np
import os
import math
from concurrent.futures import ThreadPoolExecutor

# Input and output folder paths
input_folder = r"C:\Users\Pradhyumna R Shetty\Pictures\Internship 8th sem\Frames\Frames Bypass"  # Update the path
output_folder = r"C:\Users\Pradhyumna R Shetty\Pictures\Internship 8th sem\Analog\SA1"  # Update the path

# Ensure the output folder exists, create it if not
os.makedirs(output_folder, exist_ok=True)

def process_image(filename):
    image_path = os.path.join(input_folder, filename)
    image = cv2.imread(image_path)

    if image is None:
        print(f"Error: Could not load image at {image_path}")
        return

    # Convert to grayscale
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    # Apply threshold
    _, thresh_image = cv2.threshold(gray_image, 50, 255, cv2.THRESH_BINARY)

    # Find contours
    contours, _ = cv2.findContours(thresh_image, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    # Filter contours to keep squares
    square_contours = [c for c in contours if 0.8 <= (cv2.boundingRect(c)[2] / cv2.boundingRect(c)[3]) <= 1.2]

    # Select the largest contour
    largest_contour = max(square_contours, key=cv2.contourArea) if square_contours else None

    if largest_contour is not None:
        roi_x, roi_y, roi_width, roi_height = cv2.boundingRect(largest_contour)
        roi = image[roi_y:roi_y + roi_height, roi_x:roi_x + roi_width]

        # Convert ROI to grayscale and apply Gaussian blur
        gray_roi = cv2.cvtColor(roi, cv2.COLOR_BGR2GRAY)
        blurred_image = cv2.GaussianBlur(gray_roi, (15, 15), 0)

        # Canny Edge Detection with adjusted thresholds
        edges = cv2.Canny(blurred_image, 30, 100, apertureSize=3)

        # Fixed ROI Coordinates for edges
        ex, ey, ew, eh = 67, 102, 572, 500
        edges_roi = edges[ey:ey + eh, ex:ex + ew]

        # Probabilistic Hough Line Transform with adjusted parameters
        lines = cv2.HoughLinesP(edges_roi, 1, np.pi / 180, 100, minLineLength=170, maxLineGap=10)
        
        if lines is None or len(lines) == 0:
            print(f"No lines detected in image {filename}, skipping save.")
            return

        # Select the longest line as the best detected needle
        best_line = max(lines, key=lambda l: np.linalg.norm((l[0][2] - l[0][0], l[0][3] - l[0][1]))) if lines is not None else None

        if best_line is not None:
            x1, y1, x2, y2 = best_line[0]
        else:
            print(f"No valid needle detected in {filename}.")
            return

        # Detect circular gauge
        gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
        blurred = cv2.GaussianBlur(gray, (9, 9), 2)
        circles = cv2.HoughCircles(blurred, cv2.HOUGH_GRADIENT, dp=1, minDist=50, param1=200, param2=30, minRadius=50, maxRadius=300)

        if circles is not None:
            circles = np.uint16(np.around(circles))
            circle_x, circle_y, circle_r = circles[0][0]
            cv2.circle(image, (circle_x, circle_y), circle_r, (0, 255, 0), 2)
            cv2.circle(image, (circle_x, circle_y), 2, (0, 0, 255), 3)
        else:
            print(f"No circular gauge detected in {filename}, skipping save.")
            return

        # Reference line from center to circumference (Red)
        ref_x = circle_x + circle_r
        ref_y = circle_y
        cv2.line(image, (circle_x, circle_y), (ref_x, ref_y), (0, 0, 255), 2)

        # Adjust needle coordinates
        needle_x1 = roi_x + ex + x1
        needle_y1 = roi_y + ey + y1
        needle_x2 = roi_x + ex + x2
        needle_y2 = roi_y + ey + y2

        # Draw detected needle line (Green)
        cv2.line(image, (needle_x1, needle_y1), (needle_x2, needle_y2), (0, 255, 0), 2)

        # Calculate counterclockwise angle
        angle = math.degrees(math.atan2(ref_y - circle_y, ref_x - circle_x) - math.atan2(needle_y2 - circle_y, needle_x2 - circle_x))
        if angle < 0:
            angle += 360

        # Calculate reading based on conditions
        if 180 <= angle <= 187:
            reading = 60 + (angle - 179) * (40 / 6)
        elif 162 <= angle <= 150:
            reading = 100 + (angle - 159) * (100 / 20)
        else:
            reading = None  # Undefined range

        # Display angle and reading on the image
        angle_text = f"Angle: {angle:.2f}°"
        cv2.putText(image, angle_text, (circle_x + 50, circle_y - 20), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)

        if reading is not None:
            reading_text = f"Reading: {reading:.2f}"
            cv2.putText(image, reading_text, (circle_x + 50, circle_y + 20), 
                        cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)

        # Save output image with annotations
        output_angle_path = os.path.join(output_folder, f"angle_{filename}")
        cv2.imwrite(output_angle_path, image)
        print(f"Image with angle annotation saved as {output_angle_path}")

# Process images in parallel
with ThreadPoolExecutor() as executor:
    executor.map(process_image, [filename for filename in os.listdir(input_folder) if filename.endswith((".jpg", ".png"))])
```
